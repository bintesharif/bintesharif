**1. Importing Libraries and Defining the Dictionary**

```python
from langchain.tools import tool

from langchain.agents import initialize_agent, AgentType



students = {
    "PIAIC00001": {"Name": "Ali Khan",    "Class": "10th", "Shift": "Morning"},
    "PIAIC00002": {"Name": "Sara Ahmed",  "Class": "9th",  "Shift": "Evening"},
    "PIAIC00003": {"Name": "Zara Sheikh", "Class": "8th",  "Shift": "Morning"},
    "PIAIC00004": {"Name": "Usman Raza",  "Class": "8th",  "Shift": "Evening"},
    "PIAIC00005": {"Name": "Aisha Malik", "Class": "7th",  "Shift": "Morning"},
    "PIAIC00006": {"Name": "Bilal Anwar", "Class": "6th",  "Shift": "Evening"},
    "PIAIC00007": {"Name": "Hina Tariq",  "Class": "5th",  "Shift": "Morning"},
    "PIAIC00008": {"Name": "Kamran Ali",  "Class": "4th",  "Shift": "Evening"},
    "PIAIC00009": {"Name": "Nida Aslam",  "Class": "3rd",  "Shift": "Morning"},
    "PIAIC00000": {"Name": "Omar Farooq", "Class": "2nd",  "Shift": "Evening"},
}
```

** Creating a Tool**

```python


@tool
def get_student_info(student_id: str) -> str:
    """
    Yeh function student ki ID leta hai (string)
    aur students dictionary se detail nikal kar wapas deta hai.
    """
    if student_id in students:
        data = students[student_id]
        return (
            f"Student ID: {student_id}\n"
            f"Name: {data['Name']}\n"
            f"Class: {data['Class']}\n"
            f"Shift: {data['Shift']}"
        )
    else:
        return f"Student with ID {student_id} not found!"
```

**Initializing Tools and Agent**


```python
tools_list = [get_student_info]

agent = initialize_agent(
    tools=tools_list,
    llm=llm,  # Yahaan hum apna defined LLM denge
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)
```

**Asking the Agent for Student Info**

```python
response = agent.run("Give me details for PIAIC00004.")
print(response)
```

```python
# Query the agent
response2 = agent.run("Check student ID PIAIC00004")
print(response2)
```
